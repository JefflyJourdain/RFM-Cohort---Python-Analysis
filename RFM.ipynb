```python
#Import libraries 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns 
from datetime import datetime, timedelta
```

```python
#load and take a look 
df = pd.read_csv("C:\\Users\\mj\\Documents\\DA stuff\\Datasets\\archive\\online_retail_II.csv")
df.head()
df.info()
```

```python
df['Invoice'] =  df['Invoice'].str.replace(r"a-zA-Z","", regex=True)
df['Invoice'].str.contains(r"a-zA-Z","", regex=True)

df = df.dropna(subset=['Customer ID']).copy()
df.isnull().sum()
```

here we will isolate the transaction product codes from the operational non-standard product codes, now, there are non standard codes that still represent a transaction, like D(discounts) and cruk(a donation transaction) So i have to distinguish which are worth keeping for our cohort analysis. 

```python
non_standard_codes = df[df['StockCode'].str.match(pat= '([a-z])', case=False) ]
unique_codes = non_standard_codes['StockCode'].unique()


customer_codes = ['POST','DOT', 'C2', 'D', 'M','SP1002','CRUK']
business_codes = ['TEST001', 'TEST002', 'BANK CHARGES', 'ADJUST2','DOT','ADJUST' ]

cleaned_codes = df[~df['StockCode'].isin(business_codes)] 

df = cleaned_codes.drop_duplicates()
df.duplicated().sum()
df.head()
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['Revenue'] = df['Quantity'] * df['Price']
df.head()

reference_date = df['InvoiceDate'].max() + timedelta(days=1)
reference_date

rfm = df.groupby('Customer ID').agg({
    'InvoiceDate': lambda x: (reference_date - x.max()).days,
    'Invoice': 'count',
    'Revenue': 'sum'
})

```

```python
rfm.rename( columns= {'InvoiceDate': 'Recency', 'Invoice':'Frecuency','Revenue':'Value'})
rfm.head()
```
